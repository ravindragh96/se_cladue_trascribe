```python
import os
import requests
from azure.identity import DefaultAzureCredential
```

```python
# ============================================================
# CONFIGURATION
# ============================================================

AZURE_OPENAI_ENDPOINT = (
    "https://service-experts-demo-resource.openai.azure.com"
)

MODEL_NAME = "gpt-4o-transcribe-diarize"

AUDIO_FILE = r"C:\Users\007416\Downloads\sample.mp3"
```

```python
# ============================================================
# GET ENTRA ID TOKEN
# ============================================================

credential = DefaultAzureCredential()

token = credential.get_token(
    "https://cognitiveservices.azure.com/.default"
).token
```

```python
# ============================================================
# TRANSCRIPTION ENDPOINT
# ============================================================

url = (
    f"{AZURE_OPENAI_ENDPOINT}"
    f"/openai/v1/audio/transcriptions"
)
```

```python
# ============================================================
# HEADERS
# ============================================================

headers = {
    "Authorization": f"Bearer {token}"
}
```

```python
# ============================================================
# FORM DATA
# ============================================================

data = {
    "model": MODEL_NAME,
    "response_format": "diarized_json",
    "language": "en",

    # Required for diarization
    "chunking_strategy": "auto"
}
```

```python
# ============================================================
# AUDIO FILE
# ============================================================

with open(AUDIO_FILE, "rb") as audio_file:

    files = {
        "file": (
            os.path.basename(AUDIO_FILE),
            audio_file,
            "audio/mpeg"
        )
    }

    response = requests.post(
        url,
        headers=headers,
        data=data,
        files=files,
        timeout=600
    )
```

```python
# ============================================================
# RESPONSE
# ============================================================

print("\nStatus Code:", response.status_code)

print("\nResponse:")

try:
    print(response.json())

except Exception:
    print(response.text)
```

```python
import os
import json
import requests

from azure.identity import DefaultAzureCredential


# ============================================================
# CONFIG
# ============================================================

AUDIO_FILE = r"C:\Users\007416\Downloads\sample.mp3"

ENDPOINT = (
    "https://service-experts-demo-resource.openai.azure.com"
    "/openai/deployments/"
    "gpt-4o-transcribe-diarize-2"
    "/audio/transcriptions"
)

API_VERSION = "2025-03-01-preview"


# ============================================================
# GET ENTRA ID TOKEN
# ============================================================

credential = DefaultAzureCredential()

token = credential.get_token(
    "https://cognitiveservices.azure.com/.default"
).token


# ============================================================
# URL
# ============================================================

url = (
    f"{ENDPOINT}"
    f"?api-version={API_VERSION}"
)


print("Calling:")
print(url)


# ============================================================
# HEADERS
# ============================================================

headers = {
    "Authorization": f"Bearer {token}"
}


# ============================================================
# FORM DATA
# ============================================================

data = {
    "model": "gpt-4o-transcribe-diarize",
    "language": "en",
    "response_format": "diarized_json",
    "chunking_strategy": "auto"
}


# ============================================================
# AUDIO
# ============================================================

with open(AUDIO_FILE, "rb") as audio_file:

    files = {
        "file": (
            os.path.basename(AUDIO_FILE),
            audio_file,
            "audio/mpeg"
        )
    }

    response = requests.post(
        url,
        headers=headers,
        data=data,
        files=files,
        timeout=600
    )


# ============================================================
# RESPONSE
# ============================================================

print("\nStatus Code:", response.status_code)

print("\nResponse:")

try:

    result = response.json()

    print(
        json.dumps(
            result,
            indent=2,
            ensure_ascii=False
        )
    )

except Exception:

    print(response.text)
```
